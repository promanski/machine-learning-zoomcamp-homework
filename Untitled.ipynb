```python
a = np.random.randint(low=0, high=1000, size=(3, 3))
b = np.random.randint(low=0, high=1000, size=(3, 3))

def vector_vector_multiplication(a, b):
    assert a.shape[0] == b.shape[0], "Vectors must be the same length"
    n = a.shape[0]
    result = 0.0
    for i in range(n):
        result += a[i] * b[i]
    return result

vector_vector_multiplication(a[0], b[0])
```
